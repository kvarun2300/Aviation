```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df[df["ORIGIN_AIRPORT"].str.len() == 3]
df = df[df["DESTINATION_AIRPORT"].str.len() == 3]
print("Remaining rows:", len(df))
```

```python
cancelled_flights = df['CANCELLED'].sum()
print("cancelled_flights:", cancelled_flights)
```

Top Airlines with Highest Cancellation Rate

```python
cancel_rate_airline = round(
    df.groupby("AIRLINE")["CANCELLED"]
    .mean()
    .sort_values(ascending=False) * 100,2
)

print(cancel_rate_airline)
```

```python
df.groupby("AIRLINE")['CANCELLED'].value_counts()
```

Airports with Highest Average Departure Delay

```python
dep_delay_airport = (
    df.groupby("ORIGIN_AIRPORT")["DEPARTURE_DELAY"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(dep_delay_airport)
```

Airports with Highest Arrival Delay

```python
arr_delay_airport = (
    df.groupby("DESTINATION_AIRPORT")["ARRIVAL_DELAY"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(arr_delay_airport)
```

Most Common Flight Routes

```python
df["ROUTE"] = df["ORIGIN_AIRPORT"] + "-" + df["DESTINATION_AIRPORT"]
top_routes = df["ROUTE"].value_counts().head(10)
print(top_routes)
```

Routes with Highest Average Delay

```python
route_delay = (
    df.groupby("ROUTE")["ARRIVAL_DELAY"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(route_delay)
```

Which Day of Week Has Most Delays

```python
delay_by_day = (
    df.groupby("DAY_OF_WEEK")["ARRIVAL_DELAY"]
    .mean()
)

print(delay_by_day)
```

Month with Highest Cancellation Rate

```python
cancel_month = (
    df.groupby("MONTH")["CANCELLED"]
    .mean() * 100
)

print(cancel_month)
```

Which Delay Type Contributes Most

```python
delay_types = [
"AIR_SYSTEM_DELAY",
"WEATHER_DELAY",
"AIRLINE_DELAY",
"SECURITY_DELAY",
"LATE_AIRCRAFT_DELAY"
]

df[delay_types].sum().sort_values(ascending=False)
```
